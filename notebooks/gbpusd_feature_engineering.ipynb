# GBP/USD Feature Engineering

```python
import pandas as pd
import numpy as np
```

## Load Data

```python
# Load dtype map
dtype_map = pd.read_csv("../data/cleaned/gbpusd/v1_11-09-25/gbpusd_cleaned_dtypes_as_of_110925.csv").to_dict()

# Convert dtype strings to actual Python types
dtype_map = {
    col: eval(dtype_str) if "float" in dtype_str or "int" in dtype_str else "object"
    for col, dtype_str in dtype_map.items()
}

# Load cleaned data with correct types
gbpusd_cleaned = pd.read_csv("../data/cleaned/gbpusd/v1_11-09-25/gbpusd_cleaned_as_of_110925.csv", dtype=dtype_map, parse_dates=[0], index_col=0)
```

```python
gbpusd_cleaned.head()
```

## GBP/USD Features

```python
gbpused_features = gbpusd_cleaned.copy()
```

### Returns

```python
gbpused_features["simple_return"] = gbpused_features["gbp_usd_noon_rate"].pct_change(fill_method=None)
```

```python
gbpused_features["log_return"] = np.log(gbpused_features["simple_return"] + 1)
```

```python
gbpused_features.head()
```

## Export

```python
gbpused_features.to_csv("../data/features/gbpusd/v1_11-09-25/gbpusd_features_as_of_110925.csv")
gbpused_features.dtypes.to_csv("../data/features/gbpusd/v1_11-09-25/gbpusd_features_dtypes_as_of_110925.csv")
```
